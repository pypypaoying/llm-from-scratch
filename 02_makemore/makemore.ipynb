```python
words=open("names.txt",'r').read().splitlines()
```

```python
import torch
N = torch.zeros((27,27),dtype=torch.int32)
```

```python
import string
stoi = {c: i for i, c in enumerate(string.ascii_lowercase,start=1)}
stoi['.']=0
itos={i:s for s,i in stoi.items()}
```

```python
for w in words:
    chs=["."]+list(w)+ ["."]
    for ch1,ch2 in zip(chs,chs[1:]):
        N[stoi[ch1],stoi[ch2]]+=1
```

```python
g = torch.Generator()
g.manual_seed(42)
```

```python
P= N.float()
P= P / P.sum(1,keepdims=True)
P.shape
```

```python
for i in range(50):   
    out=[]
    idx=0
    while True:
        q= P[idx]
        idx = torch.multinomial(q, num_samples=1,replacement=True, generator=g).item()
        out.append(itos[idx])
        if idx==0:
            break
    print("".join(out))
```

```python
loss=0.0
num=0.0
for w in words:
    chs=["."]+list(w)+ ["."]
    for ch1,ch2 in zip(chs,chs[1:]):
        idx1=stoi[ch1]
        idx2=stoi[ch2]
        prob = P[idx1,idx2]
        logitprob= torch.log(prob)
        nll = -logitprob
        loss += nll
        num += 1
        
print(f"{loss/num}")
```

```python
xs=[]
ys=[]
num=0
for w in words:
    chs=["."]+list(w)+ ["."]
    for ch1,ch2 in zip(chs,chs[1:]):
        idx1=stoi[ch1]
        idx2=stoi[ch2]
        xs.append(idx1)
        ys.append(idx2)
        
xs=torch.tensor(xs)
ys=torch.tensor(ys)
num=xs.nelement()
num
```

```python
ys.shape
```

```python
import torch.nn.functional as F
W=torch.randn(27,27,requires_grad=True)
xenc=F.one_hot(xs,num_classes=27).float()
```

```python
for i in range(500):
    #forward pass
    logit = xenc@W
    expit = torch.exp(logit) 
    prob = expit/expit.sum(1,keepdims=True)
    loss = -prob[torch.arange(num),ys].log().mean()

    #backward pass
    W.grad = None
    loss.backward()

    #update
    W.data += -20* W.grad

    if i%100==0:
        print(f"第{i}次迭代 loss: {loss.item()}")
    
```

```python
for i in range(5):   
    out=[]
    idx=0
    while True:
        xenc=F.one_hot(torch.tensor([idx]),num_classes=27).float()
        logit = xenc@W
        expit = torch.exp(logit) 
        prob = expit/expit.sum(1,keepdims=True)
        idx = torch.multinomial(prob, num_samples=1,replacement=True, generator=g).item()
        out.append(itos[idx])
        if idx==0:
            break
    print("".join(out))
    
```
