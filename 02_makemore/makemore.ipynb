```python
words=open("names.txt",'r').read().splitlines()
```

```python
import torch
N = torch.zeros((27,27),dtype=torch.int32)
```

```python
import string
stoi = {c: i for i, c in enumerate(string.ascii_lowercase,start=1)}
stoi['.']=0
itos={i:s for s,i in stoi.items()}
```

```python
for w in words:
    chs=["."]+list(w)+ ["."]
    for ch1,ch2 in zip(chs,chs[1:]):
        N[stoi[ch1],stoi[ch2]]+=1
```

```python
g = torch.Generator()
g.manual_seed(42)
```

```python
for i in range(50):   
    out=[]
    idx=0
    while True:
        p = N[idx].float()
        q = p / p.sum()
        idx = torch.multinomial(q, num_samples=1,replacement=True, generator=g).item()
        out.append(itos[idx])
        if idx==0:
            break
    print("".join(out))
```
